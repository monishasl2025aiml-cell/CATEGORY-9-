```python

# 1.Bayesian Reasoning - Medical Diagnosis

p_disease = 0.01
p_positive_disease = 0.95
p_positive_no_disease = 0.05

# Bayes' Theorem
p_positive = (p_positive_disease * p_disease +
              p_positive_no_disease * (1 - p_disease))

p_disease_positive = (
    p_positive_disease * p_disease
) / p_positive

print("Probability of Disease:",
      round(p_disease_positive, 4))

if p_disease_positive > 0.5:
    print("Diagnosis: High Risk")
else:
    print("Diagnosis: Low Risk")
```

```python

# 2.Fuzzy Inference System - Autonomous Vehicle

def fuzzy_speed(distance):
    if distance <= 20:
        return "Slow"
    elif distance <= 50:
        return "Medium"
    else:
        return "Fast"

distance = float(input("Enter obstacle distance: "))

speed = fuzzy_speed(distance)

print("Obstacle Distance:", distance)
print("Recommended Speed:", speed)
```
